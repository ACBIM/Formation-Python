<h1> Installation de Python </h1>

<p> Il existe divers moyens d'installer le langage Python sur un ordinateur. Nous nous contenterons ici de l'installer depuis le site officiel : 

https://www.python.org/downloads/windows/

</p>

<p> Comme pour n'importe quel logiciel, Python est fréquemment entretenu, et il en sort de nouvelles versions à mesure que ses contributeurs y améliorent les fonctionnalités ou corrigent des défauts. Il est recommandé de télécharger une version stable du langage afin de s'épargner de potentiels bugs imprévus : ces derniers peuvent causer des dégâts matériels importants et constituent en outre des failles de sécurité.</p>

<center><img src="Images/Stable.png"/> </center>

<p> Nous utiliserons ici  "Windows installer" pour la version 3.14.2. Notez que cette version est disponible pour diverses architectures de processeur. Si vous ne connaissez pas l'architecture de votre machine, ouvrez "Informations Système" (Windows uniquement). </p>

<p> Une fois l'installer téléchargé, vous pouvez l'ouvrir depuis le dossier téléchargements. Optez pour une installation manuelle et prenez garde à bien cocher l'option "Add to PATH". Cela permettra à Windows de considérer Python comme une "variable environnement", ce qui signifie simplement qu'il gardera en mémoire le chemin vers le fichier exécutable de la version de Python installée. Grossièrement, cela permet à Windows de "trouver" Python plus facilement. </p>

<center><img src = "Images\InstallationPythonPATH.png"></center>

<center><img src = "Images\pip.png"></center>

<p> Cliquez sur Custom Install et vérifiez si au moins "py launcher" "pip" et "Documentation" sont cochés. Vous pouvez choisir d'installer le langage pour l'ensemble des utilisateurs ou non.

<center><img src = "Images\env.png"></center>

<center><img src = "Images\load.png"></center>

<center><img src = "Images\success.png"></center>

<p> L'installation de Python est terminée. Pour vérifier que Windows détecte bien le programme, cherchez-le depuis la barre de recherche Windows ou ouvrez l'invité de commandes (raccourci : cmd dans la barre de recherche) et tapez la commande :

python --version

Cela doit vous afficher le nom de la version de Python que vous avez téléchargée. Le cas échéant, cela signifie que Python aura été correctement installé sur votre machine.
 </p>
 <p> Vous pouvez dès à présent utiliser le langage. Essayez de le lancer depuis l'invité de commandes en tapant simplement "python". </p>

<center><img src = "Images\Verif.png"></center>

<center><img src = "Images\ConsolePython.png"></center>

<p> Les trois flèches indiquent que vous vous trouvez désormais dans la console python. Essayez d'écrire une expression mathématique. Pour sortir de la console, tapez :

 exit()
  
 </p>

<h1> Installation d'un IDE (Environnement de Développement Intégré) </h1>

<p> La console c'est bien beau, mais ça n'est définitivement pas très confortable comme outil de travail. </p>

<p> Nous allons dès lors nous atteler à l'installation d'un IDE ou Environnement de Développment Intégré, un outil très prisé des développeurs et qui nous facilitera grandement la tâche lors de l'écriture de nos futurs prorgammes. </p>

Sachez cependant que travailler sur un IDE ne constitue absolument pas un prérequis pour travailler en Python. Il constitue simplement un confort de travail non négligeable. L'IDE que nous utiliserons ici figure parmi les plus utilisés. Il s'agit de VisualStudioCode, publié par Microsoft. Malgré cela, rien ne vous empêche de télécharger l'IDE de votre choix.

<p>Voici une liste non exhaustive de liens vers des IDE : 

https://code.visualstudio.com/download VSCode (Microsoft), que nous utiliserons ici.

https://www.jetbrains.com/fr-fr/pycharm/download/?section=windows PyCharm (JetBrains), qui propose des fonctionnalités intéressantes d'auto-complétion et d'assistance par IA.

https://www.spyder-ide.org/ Spyder (Anaconda), qui a l'avantage d'être open-source

https://thonny.org/ Thonny, un IDE très léger spécialement conçu pour les débutants. Idéal pour s'entraîner !

https://www.vim.org/download.php VIM, un éditeur de texte minimaliste, à l'ancienne. Très léger et puissant. Pour fous furieux uniquement.

 Il existe aussi de nouveaux IDE "boostés à l'IA", comme Cursor ou Windsurf.

</p>
