<h2> Les opérateurs </h2>
<p> Tout comme en mathématiques, les langages de programmation font usage d'opérateurs : </p>

```python
#L'addition, la soustraction et la multiplication : 
a = 5
b = 10
c = 26
resultat = (a + b)*a - c
print("Résultat de l'opération : ", resultat)

#L'égalité est notée "==". Ne pas confondre avec l'opérateur d'affectation !!!!
comparaison = a == b
print("Les deux variables sont-elles égales ? ", comparaison)

#De même, l'inéquation est notée "!=", signifiant "différent de".
comparaison = a != b
print("Les deux variables sont-elles différentes ?", comparaison)

#On trouve également "<", "<=", ">=", ">", en tant qu'opérateurs de comparaison : ces derniers ont la même signification qu'en mathématiques

#L'opérateur modulo, noté "%", renvoie le reste de la division euclidienne. Cette dernière est notée "/".
reste1 = b%a
reste2 = c%a
print("Reste de la division de b par a :", reste1)
print("Reste de la division de c par a :", reste2)

#Il existe aussi des fonctions pour les opérations mathématiques plus poussées : 
#pow() prend en paramètre la base et l'exponant (respectivement) et donne le résultat de cette opération. 
print(pow(b, a))

#Un module nommé "math" permet d'utiliser des fonctions mathématiques plus spécifiques.
import math 
#Ici, on utilise un alias pour math, que l'on a nommé m.
#La fonction ci-dessous calcule par exemple le cosinus d'un nombre. On voit par la syntaxe qu'elle provient du module math.
print(math.cos(0))
```

<h2> Les exceptions </h2>

```python
#On demande un nombre à l'utilisateur
nombre = input("Entrez un nombre : ")

#On définit un booléen qui évalue si le nombre choisi par l'utilisateur est pair ou non, c'est à dire si 
# le reste de la division de ce nombre par 2 est égal à 0 ou pas. input() renvoyant une chaîne de caractères, il
# nous faut effectuer une conversion de la variable "nombre" en nombre entier à l'aide de la fonction int().
estPair = int(nombre) % 2 == 0

#On affiche le résultat.
print(estPair)
```

<p> Analysez le code ci-dessus, qui constitue au passage une solution possible au problème précédent. L'expérience nous montre que rien ne nous oblige à entrer un nombre au moment où le programme exécute la fonction input().
En effet, l'utilisateur peut tout à fait entrer une chaîne de caractères quelconque. Ne pouvant effectuer l'opération voulue sur une chaîne de caractère, le programme renvoie alors une erreur.
Afin de rendre cette erreur plus explicite pour l'utilisateur final, il nous faut recourir à ce qu'on appelle une exception. </p>

```python
#La syntaxe d'une exception se présente comme suit. Ici, nous souhaitons renvoyer
#un message à l'utilisateur dans le cas où ce dernier entrerait autre chose qu'un nombre entier.

nombre = input("Entrez un nombre : ")
try:
    estPair = int(nombre) % 2 == 0
    print(estPair)
except:
    print("La conversion du nombre s'est mal passée. Réessayez.")



    
```

<h1>Les blocs conditionnels</h1>
<p>Un bloc conditionnel est un bloc qui ne s'exécute que si une condition est vérifiée.</p>

```python
#exemple :
nombre1 = 33
nombre2 = 33

if nombre1 == nombre2 :
    print(nombre1, "==", nombre2)
else : 
    print(nombre1, "!=", nombre2)
```

<p>Notons qu'il est envisageable de tester davantage de solutions en utilisant elif ("sinon si") :</p> 

```python
nombre1 = "Test"
nombre2 = "Test2"

if nombre1 < nombre2:
    print(nombre1, "<", nombre2)
elif nombre1 > nombre2:
    print(nombre1, ">", nombre2)
else:
    print(nombre1, "==", nombre2)
```

<p> La syntaxe de Python permet aussi d'écrire des conditions à la manière des mathématiciens : </p>

```python
nombre1 = 10
nombre2 = 24

if 0 < nombre1 < 11 : 
    print("Le nombre ", nombre1, "est bien entre 1 et 10")

if 0 <= nombre2 <= 10 : 
    print("Le nombre ", nombre2, "est bien entre 1 et 10")
```

<h1>Blocs itératifs</h1>
<p>Le bloc itératif est le dernier bloc d'importance qu'il faut savoir maîtriser. Il s'agit ici de déterminer si un bloc d'instructions doit être répété ou non. Le bloc en question pourra être ainsi répété de 0 à n fois. </p>

```python
#exemple : 
nombre = input("Saisissez un nombre entre 1 et 10 :")

#On gère de prime abord les exceptions liées à la fonction input :
#Il s'agit de convertir le nombre entré en type int, la fonction input ne délivrant que le type str.

try:
    nombre = int(nombre)
except:
    nombre = 0
#La dernière ligne substitue 0 à la valeur stockée dans la variable nombre. Cela nous permet de ne considérer que les entrées valides de l'utilisateur, à savoir un nombre entre 1 et 10.

while not 1 <= nombre <=10:
    nombre = input("Saisissez un nombre entre 1 et 10 :")
    try:
        nombre = int(nombre)
    except:
        nombre = 0

print("Le nombre", nombre, "est bien inclus entre 1 et 10.")
```

<p> Notez que le code ci-dessus est redondant : il pourrait être davantage optimisé. Nous pouvons remédier à cela en entrant dans une boucle infinie :  </p>

```python
while True : 
    #On entre dans la boucle infinie : l'assertion est toujours vraie.
    nombre = input("Saisissez un nombre entre 1 et 10 : ")
    try:
        nombre = int(nombre)
    except:
        pass
    else:
        if 1 <= nombre <= 10:
            #La condition est remplie, on peut quitter la boucle infinie avec break
            break

print("Le nombre", nombre, "est bien inclus entre 1 et 10.")
```
