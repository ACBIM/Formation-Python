<h1> Formatage simple d'une chaîne de caractères </h1>

<p>Le formatage d'une chaîne de caractères s'effectue à l'aide de l'opérateur modulo % suivi de la lettre s.</p>

```python
"Veux-tu la %s %s %s ?" % ("pilule", "bleue", "magique")
```

<p> On peut rendre cette expression plus souple à l'aide d'un type de données appelé dictionnaire. Un dictionnaire rattache une "clé" à sa "valeur", de la même manière qu'un vrai dictionnaire associe un mot à sa définition. Ce système est, sans surprise, appelé "clé/valeur". Voyez ci-dessous : </p>

```python
"Veux-tu la %(objet)s %(couleur)s ?" % {"objet" : "pilule", "couleur" : "bleue"}
```

<p> Ce qui peut tout aussi bien s'écrire ainsi : </p>

```python
"Veux-tu la {} {} ?".format("pilule", "bleue")
```

<p> Nous pouvons dès lors manipuler plus aisément la position des variables : </p>

```python
"Veux-tu la {1} {2} {2} ?".format("pilule", "bleue", "magique")

"Do you want the {1} {0} ?".format("pill", "blue")
```

<p> De même qu'il est possible de nommer les arguments. </p>

```python
"Veux-tu la {objet} {couleur} ?".format(objet="pilule", couleur="bleue")
```

<p>2. Bloc itératif avec la boucle for, méthode range() : </p>
<p> L'utilisation de la boucle for est similaire à while tel que mentionné dans le chapitre "Les bases". Seulement, l'usage de for permet le contrôle sur le nombre d'itérations désiré, là où while répète les instructions autant de fois qu'une condition est remplie ou non. </p>

```python
for i in range(0, 10):
    print(i)


#Pour insérer un espace : 
print("\n")

for i in range(0, 11, 2):
    print(i)
```

<h1>Les listes en Python (arrays)</h1>
<p> 1. Les listes figurent parmi ce qu'on appelle les collections ou arrays en Python. Ce sont des objets particuliers qui permettent de stocker - vous l'aurez deviné - des collections de données : des nombres entiers, des chaînes de caractères, des booléens, d'autres listes (oui, il est possible de générer une liste de listes)... En bref, une liste peut stocker tout type de donnée prédéfinie par le langage. Elle est indexée de 0 à n, où n est la longeur de la liste. Une liste est modifiable, ses éléments peuvent être supprimés ou interchangés, et l'on peut naturellement y ajouter de nouvelles données.  Voyons comment manipuler ce type d'objet : </p>

```python
#Une liste est définie avec des crochets ou avec list().
listeClients = ["Bonaparte", "Zedong", "Attila", "Khan", "Caesar"]
listeClients2 = list(("Louis XIV", "Saladin", "Tepes", "Thémistocle"))

#Elle peut recevoir des types de données tout à fait différents : 
listeDonnees = ["Chaine", True, 20.3, 450, listeClients, ["ACDEAU", "ACBIM"]]

#On peut aussi initialiser une liste vide : 
liste_a_completer = []


#On accède possiblement aux différents éléments de la liste par leur index, précisé entre crochets suivant le nom de la variable : 
#Ici, on affiche le tout premier élément de listeClients, à savoir "Bonaparte"
print(listeClients[0])


#Pour ajouter un élément à la fin d'une liste : 
liste_a_completer.append("Néron")
#On peut aussi placer directement un élément via l'index : 
liste_a_completer[0] = "Octave"
#Cela a pour effet de substituer la valeur qui se trouvait précédemment à cet emplacement : on aura remplacé "Néron" par "Octave".

#Si l'on veut ajouter un élément à la liste sans remplacer la donnée présente à un index donné, il est possible d'utiliser la méthode insert()
liste_a_completer.insert(0, "Néron")
print(liste_a_completer)
#Ainsi "Néron" est inséré à l'index 0, et "Octave" prendra 1 pour index (décalé vers la droite dans la liste).


#Pour enlever un élément de la liste : 
listeClients.remove("Caesar")
print(listeClients)

#Pour enlever le dernier élément de la liste : 
listeClients2.pop()
print(listeClients2)

#Transformer une chaîne de caractères en liste : 
citation = "Demain dès l'aube, à l'heure où blanchit la campagne..."
#La fonction split peut prendre un paramètre purement optionnel indiquant le séparateur.
citationDecoupee = citation.split(sep=" ")
print(citationDecoupee)
      
```

<h1> Écrire, modifier un fichier en Python </h1>

<p> Nous avons vu comment manipuler des données propres au langage Python. Il faut cependant savoir que Python est capable d'agir directement sur des fichiers. Voyons comment se présente la chose. </p>

```python
#Ouvrir un fichier existant : 
#Il nous faut préciser le chemin relatif ou absolu du fichier dans le paramètre de la fonction open()
path = "C:/Users/BimP05/Documents/formationPython/notebook/Exercices/ressources/listingCoordonnees"
path2 = "C:/Users/BimP05/Documents/formationPython/notebook/Exercices/ressources/listingCoordonnees copy"
#Si l'on souhaite modifier le contenu du fichier, il faudra ajouter l'option "w" (pour "write")
with open(path, "w") as f:
    test = "Contenu du fichier numéro 1"
    f.write(test)

#Notez que le programme ci-dessus a pour effet de substituer entièrement le contenu préexistant du fichier. 
#Si l'on veut simplement ajouter du contenu à la suite de la donnée préexistante, il faut utiliser
#  l'option "a" (pour append) dans open():

with open(path2, "a") as f2:
    test = "Contenu du fichier numéro 2"
    f2.write("\n" + test)

#On peut aussi concaténer le contenu de deux fichiers dans un fichier tiers : 
#Utilisons open() avec la fonction "r" (pour "read")
fichier1 = open(path, "r")
fichier2 = open(path2, "r")

#On stocke le contenu des deux fichiers
contenuFichier1 = fichier1.read()
contenuFichier2 = fichier2.read()

#On utilise la méthode .close() afin de s'assurer que la ressource n'est plus utilisée
fichier1.close
fichier2.close

pathVersLeFichierFinal = "C:/Users/BimP05/Documents/formationPython/notebook/Exercices/ressources/listingFusionne"
#On souhaite écrire dans le fichier, on utilisera donc l'option "w"
with open(pathVersLeFichierFinal, "w") as f3:
    f3.write(contenuFichier1 + "\n" + contenuFichier2)


```
