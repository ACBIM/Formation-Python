<h1> Les classes </h1>
<p> Les classes sont des objets quelque peu particuliers. En fait, c'est à partir de classes qu'ont été "fabriqués" les différents types de données dans Python (int, float, str, list...). À savoir que la notion de classe n'est pas spécifique à ce langage. En réalité, elle s'étend à tout langage touchant de près ou de loin à ce que l'on appelle la POO ou Programmation Orientée Objet.

 C++ ou encore Java sont des exemples de langages de programmation ayant recours aux classes.
 </p>

<h1> Qu'est-ce que la POO ? </h1>
<p> La Programmation Orientée Objet n'est rien d'autre qu'un paradigme informatique, une façon d'appréhender la programmation. Cette dernière est portée sur l'interaction d'objets, qui peuvent représenter autant d'entités que l'on souhaite (un client, une voiture, un film...). </p>

<p> Pour représenter un objet en informatique, nous avons donc recours aux classes. Une classe représente en fait une catégorie d'objets : c'est en quelque sorte un moule qui va nous servir à fabriquer des "instances", c'est à dire des objets qui ont leur particularités mais qui rentrent dans la même catégorie. Par exemple, les instances "Madeleine" et "Gâteau au chocolat" pourraient découler de la classe "Pâtisserie", bien qu'elles aient des caractéristiques propres.

 Abordons cette notion de manière plus concrète en définissant notre propre classe :
 </p>

```python
#Par convention, le nom d'une classe commence toujours par une majuscule, à l'inverse des variables ou des fonctions.

class Client:
    #Cette syntaxe quelque peu particulière s'apppelle un constructeur. Retenez simplement qu'elle permet de paramétrer
    #l'objet avec les paramètres souhaités, un peu comme si l'on définissait une fonction :
    def __init__(self, nom, prenom, coordonnees, societe):
        self.nom = nom
        self.prenom = prenom
        self.coordonnees = coordonnees
        self.societe = societe


client1 = Client("Bonaparte", "Napoléon", "07 98 35 69 45", "Premier Empire")
print(client1.societe)
#la variable client1 reçoit toutes les caractéristiques de l'instance de Client que l'on a produite.
```

<p> Une classe peut également contenir ses propres méthodes : </p>

```python
#Dans le même esprit :
class Client:
    def __init__(self, nom, prenom, coordonnees, societe):
        self.nom = nom
        self.prenom = prenom
        self.coordonnees = coordonnees
        self.societe = societe

    def salutationsPersonnalisees(self):
        print("Bonjour, " + self.prenom + " " + self.nom  + "!")

#On instancie notre classe
client1 = Client("Cesar", "Jules", "07 83 56 88 24", "Pontifex Maximus")

client1.salutationsPersonnalisees()
```
